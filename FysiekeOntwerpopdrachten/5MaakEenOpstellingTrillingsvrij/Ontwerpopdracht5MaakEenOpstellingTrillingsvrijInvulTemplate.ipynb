## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
omega=5*2*np.pi
f0= omega/(4*np.pi)
print(f0)
```

```python
# Voeg hier je plot in.
frequentix=np.linspace(0,10,500)
def Ampli(f):
   return 1/(1-(f**2/f0**2))

plt.figure()
plt.loglog(frequentix,abs(Ampli(frequentix)))
plt.xlabel('Frequentie (Hz)')
plt.ylabel('Verhouding van Amplitude')
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
omega0=f0*2*np.pi
g=9.81
Delta_x=g/omega0**2
print(Delta_x*100,'cm')
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().
massa=np.linspace(0,10,100)
def veerconstante(a):
    return massa*g/(a*Delta_x)

plt.figure()
plt.plot(massa,veerconstante(1),label='dx')
plt.plot(massa,veerconstante(2),label='2 keer zo groot dx')
plt.plot(massa,veerconstante(5),label='5 keer zo groot dx')
plt.xlabel("massa")
plt.ylabel('veerconstante')
plt.legend()
plt.grid(True)
plt.show()

print((750*0.094/(7.0*10**-4))*10**-6,'GPa')
```
